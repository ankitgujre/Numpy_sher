```python
import numpy as np

# Advance Array Manipulation
```

```python
arr = np.array(range(1,21))
arr
```

```python
a = np.arange(1,11)
a
```

```python
A = np.array([[1,2],[3,4]])
print(A)
B = np.array([[5,6],[7,8]])
print(B)
```

```python
# Stacking Array
A = np.arange(1,5)
print(A)
B = np.arange(5,9)
print(B)
```

```python
vs = np.vstack((A,B))
print(vs)
np.ndim(vs)
```

```python
hs = np.hstack((A,B))
print(hs)
np.ndim(hs)
```

```python
col_st = np.column_stack((A,B))
print(col_st)
```

```python
np.ndim(col_st)
```
