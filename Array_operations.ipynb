```python
import numpy as np

# Arithmetic Operations
```

```python
a1 = np.array([1,2,3,4,5])
a2 = np.array([6,7,8,9,10])
print(a1)
print(a2)
```

```python
arrAdd = a1 + a2
print(arrAdd)
```

```python
arrSub = a1 - a2
print(arrSub)
```

```python
arrMul = a1 * a2
print(arrMul)
```

```python
arrDivide = a1/a2
print(arrDivide)
```

```python
arrFloor = a1//a2
print(arrFloor)
```

```python
arrExpon = a1 ** a2
print(arrExpon)
```

```python
# Broad Casting

l = [10,20,30,40,50]
arr = np.array(l)
print(arr)
```

```python
arr + 10
```

```python
arr2 = np.arange(1,26).reshape(5,5)
print(arr2)
```

```python
arr2 +10
```

```python
# Deep and Shallow copy

arr = np.arange(1,21)
arr
```

```python
slice = arr[:5]
slice = slice*10
print(slice)
```

```python
# Matrix Operations

A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])
print(A)
print()
print(B)
```

```python
A @ B
```

```python
np.dot(A,B)
```
